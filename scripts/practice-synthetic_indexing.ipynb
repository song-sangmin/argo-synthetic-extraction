```python
import ast
yr = 2014
yrcol = f'{yr}_records'
# data_path = '../../ktchris/Synthetic_Argo_Dataset/'
# synargo_4hourly = xr.open_dataset(data_path + 'Synthetic_Argo_Data_'+ str(yr) + '.nc')

print('Retrieving float data for year ' + str(yr) + ': ', datetime.now().strftime('%Y%m%d %H:%M'))

# Retrieve floats that have data for this given year
yearINDEX = deployINDEX[~deployINDEX[yrcol].isna()]
print('Processing ' + str(len(yearINDEX)) + ' valid floats...')

# Iterate over each float using that years 
fnumDatasets = []
for fnum in yearINDEX.index:
    record_list_4hourly = ast.literal_eval(yearINDEX.loc[fnum][yrcol])
    record_list = record_list_4hourly[::int(10*24/4)]

    nParticles_index = int(yearINDEX.loc[fnum]['nParticles_index'])
    floatDS = synargo_4hourly.isel(record=record_list, nParticles = nParticles_index)
    
    fnum_val = int(floatDS['fnum'].values.tolist())
    fnumDatasets.append(floatDS.assign_coords(fnum=fnum_val))

annualDS = xr.concat(fnumDatasets, dim='fnum')
annualDS.to_netcdf('./synArgo_10day_annual/synArgo_realworld_matches_10day_yr' + str(yr) + '.nc')
print('Finished retrieving year ' + str(yr) + ': ', datetime.now().strftime('%Y%m%d %H:%M'))
```
